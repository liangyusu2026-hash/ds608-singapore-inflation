```python
# Import necessary libraries
import pandas as pd
import plotly.express as px

df = pd.read_csv("outputs/tables/inflation_by_category.csv")

# inspect all the series in the dataset
series_list = df["series"].unique()
print(series_list)  

df["date"] = pd.to_datetime(df["date"])

print(df["date"].max())
```

```python
contributions = pd.read_csv(
    "outputs/tables/contributions_monthly.csv"
)

contributions.head()
```

```python
# Select September 2022 and rank categories
peak_month = contributions[
    contributions["date"] == "2022-09-01"
]

ranked = peak_month.sort_values(
    "contribution_pp",
    ascending=True
)

# Create a horizontal bar chart
fig = px.bar(
    ranked,
    x="contribution_pp",
    y="series",
    orientation="h",
    title="Contributions to Singapore Inflation — September 2022",
    labels={
        "contribution_pp": "Contribution (percentage points)",
        "series": "Expenditure Category"
    }
)

fig.show()
```

```python
# Extract Food series data
food = df[df["series"] == "Food"]
display(food.head())

# Extract Transport series data
transport = df[df["series"] == "Transport"]
transport.head()
```

```python
# Create a line chart for Food CPI over time

fig_food_cpi = px.line(
    data_frame=food,
    x="date",
    y="cpi",
    title="Singapore Food CPI Historical Trend (Base Year 2024=100)"
)

fig_food_cpi.show()

# Create a line chart for Food Inflation (YoY % Change) over time
fig_food_inflation = px.line(
    data_frame=food,
    x="date",
    y="yoy_pct",
    title="Singapore Food Inflation (YoY % Change)"
)
fig_food_inflation.show()   

# Create a line chart for Transport CPI over time
fig_transport_cpi = px.line(
    data_frame=transport,
    x="date",
    y="cpi",
    title="Singapore Transport CPI Historical Trend (Base Year 2024=100)"
)
fig_transport_cpi.show()

# Create a line chart for Transport Inflation (YoY % Change) over time
fig_transport_inflation = px.line(
    data_frame=transport,
    x="date",
    y="yoy_pct",
    title="Singapore Transport Inflation (YoY % Change)"
)
fig_transport_inflation.show()   
```
